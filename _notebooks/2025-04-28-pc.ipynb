```python
# List to store reservations (data abstraction) to represent a collection of data that is stored and used to manage program complexity and help fulfill the program’s purpose.
# Throughout the program, the list is used to Stpre reservation entries (each is a dictionary), check for time conflicts (iteration), and display current reservations.
reservations = []

# Procedure to add a reservation.

# Procedure's name is add_reservation.
# The return type (str) returns a confirmation or error message.
# One or more parameters (name: str , size: int, time: str).
# Procedure manages restaurant reservations.
def add_reservation(name: str, size: int, time: str) -> str:
    # Selection: if statement choose different paths based on conditions.
    if not name or size <= 0 or not time:
        return "Error: Please enter a valid name, party size, and time."

    # Selection and iteration: check for time conflicts
    # Loop through the reservations list to check for time conflicts.

    for res in reservations:
        if res["time"] == time:
            return f"Error: The time slot at {time} is already booked."

    # Add reservation (manages complexity using list of dictionaries)
    reservations.append({"name": name, "size": size, "time": time})
    return f"Reservation confirmed for {name} at {time} for {size} people."
# Sequencing Breakdown: Input validation comes first, so bad data is rejected early. Time conflict checking comes next, only if the inputs are valid. Appending the reservation happens only if there’s no conflict. Returning a message wraps it up.

# Procedure to display all current reservations (text output)
def display_reservations():
    if not reservations:
        print("\nNo reservations yet.")
    else:
        print("\nCurrent Reservations:") # Textual output based on the stored data in list, showing visual/textual feedback of current reservations.
        for res in reservations:
            print(f"- {res['name']} for {res['size']} people at {res['time']}")

# Main loop (input/output logic)
# In the main() function, it takes input from the user using: input("Enter your name: "), input("Enter party size: "), and input("Enter reservation time (e.g., 7:00 PM): ")
# User actions (like selecting menu options and typing information) are what trigger events like adding a reservation or displaying existing ones.
def main():
    print("Welcome to the San Diego Restaurant Reservation System!")
    while True:
        print("\nReservation Options:")
        print("1. Make a reservation")
        print("2. View all reservations")
        print("3. Exit")
        choice = input("Choose an option (1-3): ")

        if choice == "1":
            name = input("Enter your name: ")
            try:
                size = int(input("Enter party size: "))
            except ValueError:
                print("Error: Party size must be a number.")
                continue
            time = input("Enter reservation time (e.g., 7:00 PM): ")
            message = add_reservation(name, size, time) # Call to procedure; collect users input (name, size, time) call to custom procedure (add_reservation(name, size, time)), then the result (confirmation or error message) is stored in the variable message. Finally, message is print to the user.
            print(message)  # Textual output based on input and program functionality. This prints the result of add_reservation, which could be; A confirmation: "Reservation confirmed for (name) at (time) for (size)." or an error: "Error: The time slot at (time) is already booked."
        elif choice == "2":
            display_reservations()
        elif choice == "3":
            print("Goodbye!")
            break
        else:
            print("Invalid option. Try again.")

# Run the program
main()
```
